### Task 3: Load Data from SQL Database
1. Create a SQLite database (sample.db)
2. Create a table named employees with columns:
o id

o salary
3. Insert at least 5 records.
4. Read the table into Pandas using SQL query.
5. Display the DataFrame.

```python
import sqlite3

try:
    # Connect to SQLite Database and create a cursor
    conn = sqlite3.connect('sample.db')
    cursor = conn.cursor()
    print('DB Init')

    # Execute a query to create table
    query = "Create Table IF NOT EXISTS Employee (id integer PRIMARY KEY, name varchar(50), department varchar(50), salary integer);" \
    "Insert into Employee values (1, 'Emp1', 'A', 100000); Insert into Employee values(2, 'Emp2','A', 15000); Insert into Employee values(3, 'Emp3','B', 12000); Insert into Employee values(4, 'Emp4', 'c', 15600); Insert into Employee values(5, 'Emp5', 'B', 60000);"
    cursor.executescript(query)

    # Close the cursor after use
    cursor.close()

except sqlite3.Error as error:
    print('Error occurred -', error)

finally:
    # Ensure the database connection is closed
    if conn:
        conn.close()
        print('SQLite Connection closed')
```
