# Task 2: Read file in Different Ways

```python
"""
There are 2 ways of accessing a file 
1. Using open() - file needs to be closed explicitly
2. Using with open() - file will be closed implicitly
below code demonstrates the usage of both the methods
"""

# read function - reads the complete file in 1 short as a string
f = open('sales_data.txt', 'r')
data = f.read()
print("read()")
print('Data from file: ', data, type(data))
# split the string to convert output in the list format
data = data.split()
# removing new line and covert it to int
data = list(map(lambda x: int(x), data))
print('Data after processing: ', data, type(data))
f.close()

# readline - reads only the first line of the file
f = open('sales_data.txt', 'r')
print('\nreadline()')
data1 = f.readline()
print('Data from file: ', data1, type(data1))
data1 = data1.split()
# removing new line and covert it to int
data1 = list(map(lambda x: int(x), data1))
print('Data after processing: ', data1, type(data1))

f.close()

# readlines - reads the complete file as each row as element of the list
with open('sales_data.txt', 'r') as f:
    print('\nreadlines()')
    data2 = f.readlines()
    print('Data from file: ', data2, type(data2))
    # removing new line and covert it to int
    data2 = list(map(lambda x: int(x), data2))
    print('Data after processing: ', data2, type(data2))
```

# Task 3 - Append new Sales

```python
# Append new sales to the same file 1000, 2500, 1700

# a+ will help in appending and reading the existing file with curser at the end of the file
new_sale = [1000, 2500, 1700]
with open('sales_data.txt', '+a') as f:
    for x in new_sale:
        f.write(str(x)+'\n')

# after appending reopen and print updated file
with open('sales_data.txt', '+r') as f:
    data = f.readlines()
    print('Data after the file update: ', data, type(data))
    # removing new line and covert it to int
    data = list(map(lambda x: int(x), data))
    print('Data after processing: ', data, type(data))

# print total lines after appending
    print('Total lines after the file update: ', len(data))
```

# Task 4 - Generate summary report from the file

```python
# read all file values
with open('sales_data.txt', 'r+') as f:
    data = f.readlines()
    print('Data read from file : ', data)

    # convert them to integer
    data = list(map(lambda x: int(x), data))
    print('Data converted to integer : ', data)

# calculate following
# total sale, highest sale, lowest sale, average sale
highest_sale = 0
lowest_sale = 0
avg = 0
total = 0  
counter = 0
for i in range(len(data)):
    if i == 0:
        highest_sale = data[i]
        lowest_sale = data[i]
        total = data[i]
    else:
        # comparison of data for highest value
        if data[i]>highest_sale:
            highest_sale = data[i]
        # comparison of data for lowest value
        if data[i]<lowest_sale:
            lowest_sale = data[i]
        # adding up the total data
        total += data[i]
    counter+=1
avg = int(total/counter)
print('Total Sale: ', total)
print('Highest Sale: ', highest_sale)
print('Lowest Sale: ', lowest_sale)
print('Average Sale: ', avg)
```

# Task 5 - Create product Info File (User Input)

```python
# Ask user 3 product names and their price
product = []
for i in range(1, 4):
    prod_name = input('Enter the '+ str(i)+ ' product name: ')
    prod_price = input('Enter the '+ str(i)+ ' product price: ')
    product.append(prod_name+' | '+prod_price+'\n')
print("List of products: ", product)

# write them into a products.txt ProductName | Price
with open('products.txt', 'w+') as f:
    f.writelines(product)
```

```python
# Read the file and print each line with proper formating
with open('products.txt', 'r+') as f:
    data = f.readlines()
    print('Data read from the file: ', data)

# loop over the file data for formatted output
# counter is used for refering the product number
counter = 0
for i in data:
    # split will split the data into list from ' | '
    prod = i.split(' | ')
    print('\nProduct %d name: '%(counter+1), prod[0])
    # strip will take off whitespaces or escape sequence from the start and end of the string
    print('Product %d price: '%(counter+1), prod[1].strip())
    counter+=1
```

```python
import os
# Ask the user for a filename to open
fname = input("Enter the filename to be opened <filename>.<extension>: ")
# check the validity of the filename, we can apply further constraints to check extention as well
if '.' in fname:
    # if file exisits read and print
    # print(os.path.exists(fname))
    if os.path.exists(fname):
        f = open(fname, 'r+')
        data = f.readlines()
        f.close()
        data = list(map(lambda x:x.strip() , data))
        print('Data in the file: ', data)
        for x in data: print(x)
    # else print error
    else:
        print('File not found. Please check the filename')
else:
    print('File name invalid. Please check the filename')
```

```python
# Create a dictionary
prices = {
    'Mouse':500, 'Keyboard':800, 'Monitor':7000, 'Pendrive':400, 'Camera': 5000
}
# Ask the user for a discount percentage
discount_pct = int(input("Enter the discount percentage: "))

# write discounted prices into discount_report.txt Product | original Price | Discounted Price
product = []
for k, v in prices.items():
    dp = (1 - (0.01)*discount_pct)*v
    product.append('%s | %d | %.2f\n'%(k,v,dp) )

print('Data to be written to the file: ',product)
f = open('discount_report.txt', 'w+')
f.writelines(product)
f.close()

# read the file and print
f = open('discount_report.txt', 'r+')
data = f.readlines()
f.close()
print("Data read from the file: ", data)

# loop over the file data for formatted output
# counter is used for refering the product number
counter = 0
total = 0
for i in data:
    # split will split the data into list from ' | '
    prod = i.split(' | ')
    print('\nProduct %d name: '%(counter+1), prod[0])
    print('Product %d price: '%(counter+1), prod[1])
    # strip will take off whitespaces or escape sequence from the start and end of the string
    d_pc = prod[-1].strip()
    print('Product %d discounted price: '%(counter+1), d_pc)
    total+=float(d_pc)
    counter+=1

# File summary Total Items: X Average Discounted Price: Y
avg = total/counter
print("\nTotal Items: %d\nAverage Discounted Price %.2f"%(counter, avg))
```
