# Task 1 - Discount Rules (if/ elif / else)

```python
# logic for discount
def discountCalculation(order_amount):
    order_amount = int(order_amount)
    if order_amount>=2000:
        discount = order_amount*0.15
    elif order_amount<2000 and order_amount>=1500:
        discount = order_amount*0.10
    elif order_amount<1500 and order_amount>=1000:
        discount = order_amount*0.07
    else:
        discount = 0

    amount_after_discount = order_amount-discount
    tax = amount_after_discount*0.05
    return (order_amount, discount, amount_after_discount, tax, int(tax+amount_after_discount))
```

```python
# code to read integer input from user
order_amount = input('Enter order amount - ')

if not order_amount.isdigit():
    print('Entered amount is not correct please try again!')
else:
    
    order_amount, discount, amount_after_discount, tax, final_amount = discountCalculation(order_amount)
    print('Order Amount: ',order_amount)
    print('Discount Amount: ',discount)
    print('Subtotal: ',amount_after_discount)
    print('Tax 5%: ', tax)
    print('Total amount to be paid: ', final_amount)
```

# Task 2 - Process multiple orders in the loop

```python
orders = [1200, 2500, 800, 1750, 3000]
print('Order Amount' , 'Discount', 'Final Amount', 'Tax')
order_discount = 0
total_revenue = 0
for order in orders:
    order_amount, discount, amount_after_discount, tax, final_amount = discountCalculation(order)
    print(order_amount, int(discount), int(final_amount), int(tax))
    total_revenue += amount_after_discount
    if discount>0: order_discount+=1

print('Total Revenue after discount: ', total_revenue)
print('Orders with discount > 0: ', order_discount)
```

# Task 3 - User Menu (while loop - Break / continue)

```python
def menu():
    print('Press the following for actions')
    print('1. Enter order amount to the running list')
    print('2. Show all orders and totals after discount')
    print('q Quit')
    choice = input('Enter your choice')
    return choice

choice = menu()
orders_list = []
while choice:
    if choice=='q' or choice=='Q':
        break
    elif choice in ['1', '2']:
        if choice=='1':
            order_amount = input('Enter order amount - ')
            if not order_amount.isdigit():
                print('Entered amount is not correct please try again!')
                continue
            else:
                orders_list.append(order_amount)
        elif choice=='2':
            print('Order_Amount' , 'Discount', 'Final_Amount', 'Tax')
            for order in orders_list:
                order_amount, discount, amount_after_discount, tax, final_amount = discountCalculation(order)
                print(order_amount, int(discount), int(final_amount), int(tax))
    else:
        print('Your choice is incorrect! Try again!')
    choice = menu()
    # continue

    
```

# Task 5 - Loop control with conditions (break and continue)

```python
daily =[200, 150, 0 , 400, 50, -1, 300]
total_sale = 0
for sale in daily:
    if sale<=-1:
        print('Corrupted Data')
        break
    elif sale==0:
        print('No sale on the day')
        continue
    elif sale>0:
        total_sale+=sale
        print('Current day Sale: ', sale)
        print('Total sale so far = ', total_sale)

print("Total Sale: ", total_sale)


```
