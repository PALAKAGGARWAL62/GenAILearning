# Task 1 - Basic Function : price after discount

```python
def apply_discount(price, *args):
    if args:
        discount = args[0]
    else:
        discount = 5 
    if discount>60:
        print('Discount can not be more than 60 %! Try Again')
        return price
    price_discount = (price*(100-discount))/100
    return int(price_discount)
```

```python
price = apply_discount(1000, 10)
print(price)
price = apply_discount(500)
print(price)
```

# Task 2 - Recursive Function: Factorial Utility

```python
def factorial(n):
    if n==1:
        result = 1
    elif n==0:
        result = 0
    elif n>1:
        result = n*factorial(n-1)
    else:
        result = 'Number is negative, Try again'
    return result

f = factorial(5)
print('Factorial 5 - ',f)
f = factorial(0)
print('Factorial 0 - ',f)
f = factorial(-3)
print('Factorial -3 - ',f)
```

# Task 3 - Lambda Function: GST Calculator

```python
gst = lambda price: price+(0.18*price)
print(gst(100))
gst_discount = lambda price, discount: (price*(100-discount))/100+(0.18*price)
print(gst_discount(100, 10))
```

# Task 4- Using map(): Apply GST to list of prices

```python
prices = [100, 250, 400, 1200, 50]
prices_with_gst = list(map(gst, prices))
print('Original Prices -', prices)
print('Prices with GST -', (prices_with_gst))
```

# Task 5- Using filter(): Filter Expensive Products

```python
prices = [100, 250, 400, 1200, 50, 2000, 850]
prices_gt500 = list(filter(lambda price: price>500, prices))
prices_lte500 = list(filter(lambda price: price<=500, prices))
print('Original Prices -', prices)
print('Prices > 500 -', prices_gt500)
print('Prices <= 500 -', prices_lte500)
```

# Task 6 - Combined Utility Function

```python
def process_prices(prices):
    discounted_prices = list(map(lambda price: price*0.9, prices))
    filtered_prices = list(filter(lambda price: price>300, discounted_prices))
    return discounted_prices, filtered_prices
prices =[100, 500, 900, 50, 750]
dp, fp = process_prices(prices)
print('Original Prices -', prices)
print('Discounted Prices -', dp)
print('Filtered Prices -',fp)


```

# Task 7 - Mini Problem: Menu Using Functions

```python
def add_price(prices_list, price):
    # add_lambda = lambda price, tobeadded: price+tobeadded
    # result = list(map(add_lambda(prices_list, price), prices_list))
    print('Original Prices List: ', prices_list)
    prices_list.append(price)
    print("Updated Prices List: ", prices_list)
    return prices_list

def get_average_price(prices_list):
    result = 0
    for x in prices_list: result += x 
    result /= len(prices_list)
    print('Original Prices List: ', prices_list)
    print('Average Price: ', result)
    return result

def get_max_price(prices_list):
    result = 0
    for x in prices_list:
        if x>result:
            result = x
    print('Original Prices List: ', prices_list)
    print('Maximun Price: ', result)
    return result

def menu():
    print('Press the following for actions')
    print('1. Add prices to the list')
    print('2. Average Price')
    print('3. Maximum Price')
    print('q Quit')
    choice = input('Enter your choice')
    return choice

choice = menu()
prices_list = []
while choice:
    if choice=='q' or choice=='Q':
        break
    elif choice in ['1', '2', '3']:
        if choice=='1':
            price = input('Enter price - ')
            if not price.isdigit():
                print('Entered amount is not correct please try again!')
                continue
            else:
                add_price(prices_list, int(price))
        elif choice=='2':
            get_average_price(prices_list)
        elif choice=='3':
            get_max_price(prices_list)
    else:
        print('Your choice is incorrect! Try again!')
    choice = menu()
```
