## Task 1: Pandas Series Basics

```python
'''
1. Import pandas as pd.
2. Create a Pandas Series from the list:
3. marks = [78, 85, 90, 66, 72]
4. Print:
o Series values
o Index
o Data type
5. Access:
o First element
o Last two elements
'''
import pandas as pd
marks = [78, 85, 90, 66, 72]
marks_serie = pd.Series(marks)
print("Marks Series : \n", marks_serie)
print('Index of Mark Series: ', marks_serie.index)
print('Data Type of Mark Series: ', marks_serie.dtype)
# (Getting values using array index)
print('First element: ', marks_serie.values[0])
# Another way (Getting values using series index)
print('First element: ', marks_serie[0])
# (Getting values using array index)
print('Last element: ', marks_serie.values[-1])
# Another way (Getting values using series index)
print('Last element: ', marks_serie[4])
```

## Task 2: Mathematical Operations on Series

```python
""" 
1. Add 5 grace marks to all students.
2. Subtract 2 marks from all values.
3. Multiply all marks by 1.05.
4. Divide all marks by 2.
Print results for each operation.
"""

print("Original marks: \n",marks_serie)

# 1. Add 5 grace marks to all students.
op = marks_serie+5
print("5 Grace marks added\n", op)

# 2. Subtract 2 marks from all values.
op = marks_serie-2
print("2 marks Subtracted\n", op)

# 3. Multiply all marks by 1.05.
op = marks_serie*1.05
print("Multiply all marks by 1.05\n", op)

# 4. Divide all marks by 2.
op = marks_serie/2
print("Divide all marks by 2\n", op)
```

## Task 3: Python Functionalities on Series

```python
''' 
Using the same Series:
1. Find:
o Maximum marks
o Minimum marks
o Sum of marks
o Mean marks
2. Apply a lambda function to check whether each student has passed (>= 70).
3. Count how many students passed.
'''
# o Maximum marks
print("Maximum Marks: ", marks_serie.max())
# o Minimum marks
print("Minimum Marks: ", marks_serie.min())
# o Sum of marks
print("Sum of Marks: ", marks_serie.sum())
# o Mean marks
print("Mean Marks: ", marks_serie.mean())

# 2. Apply a lambda function to check whether each student has passed (>= 70).
op = marks_serie.apply(lambda x : x>= 70)
print('Student passed (>= 70)\n', op)
# 3. Count how many students passed.
print('Students passed: ', op.sum())
```

## Task 4: Create a DataFrame

```python
'''
Create a DataFrame students:
students = {
'Name': ['Amit', 'Neha', 'Rahul', 'Sneha', 'Pooja'],
'Marks': [78, 85, 90, 66, 72],
'Subject': ['Math', 'Math', 'Science', 'Science', 'Math']
}
1. Convert it into a DataFrame.
2. Print first 3 rows.
3. Print last 2 rows.
4. Print DataFrame shape and column names.
'''
students = {
'Name': ['Amit', 'Neha', 'Rahul', 'Sneha', 'Pooja'],
'Marks': [78, 85, 90, 66, 72],
'Subject': ['Math', 'Math', 'Science', 'Science', 'Math']
}
df = pd.DataFrame(students)
print(df)
print('\nFirst 3 rows \n', df.head(3))
print('\nLast 2 rows \n',df.tail(2))
# another way
print('\nFirst 3 rows \n', df[:3])
print('\nLast 2 rows \n',df[-2:])

print('\nDataframe  information\n')
print(df.info())
print('\nDataframe  shape\n', df.shape)
print('\nDataframe  Columns\n', df.columns.values)
```

## Task 5: Important DataFrame Functions

```python
'''Using the students DataFrame:
1. Use and print output of:
o .info()
o .describe()
o .head()
o .tail()
2. Sort students by Marks in descending order.
3. Reset index after sorting. 
'''
# 1. Use and print output of:
# o .info() - infomation related to dataframe structure
print('\nDataframe Info \n', df.info())

# o .describe() - statistical measure of numerical data
print('\nDataframe describe \n', df.describe())

# o .head() - top 5 rows of df
print('\nDataframe head \n', df.head())

# o .tail() - last 5 rows of df
print('\nDataframe tail \n', df.tail())

# 2. Sort students by Marks in descending order.
sorted_df = df.sort_values(by='Marks', ascending=False)
print('\nDataframe Sorting \n', sorted_df)

# 3. Reset index after sorting. 
sorted_df = df.sort_values(by='Marks', ascending=False, ignore_index=True)
print('\nReset index after sorting \n', sorted_df)
```

## Task 6: Filtering & Conditional Selection

```python
'''
Perform the following filters:
1. Students who scored more than 75 marks.
2. Students belonging to subject Math.
3. Students who scored more than average marks.
4. Students who failed (marks < 70).
'''

# Perform the following filters:
# 1. Students who scored more than 75 marks.
print('\nStudents who scored more than 75 marks\n', df[df['Marks']>75])
# 2. Students belonging to subject Math.
print('\nStudents belonging to subject Math\n', df[df['Subject']=='Math'])
# 3. Students who scored more than average marks.
avg_marks = df['Marks'].mean()
print("Average Marks: ", avg_marks)
print('\nStudents who scored more than average marks\n', df[df['Marks']>avg_marks])
# 4. Students who failed (marks < 70).
print('\nStudents who failed (marks < 70)\n', df[df['Marks']<70])
```

## Task 7: Grouping & Basic Analysis

```python
''' 
1. Find average marks per subject using groupby().
2. Count number of students per subject.
3. Find maximum marks per subject.
'''

# 1. Find average marks per subject using groupby().
print('\nAverage marks per subject using groupby()\n', df.groupby(by = 'Subject')['Marks'].mean())
# 2. Count number of students per subject.
print('\nCount number of students per subject\n', df.groupby(by = 'Subject')['Subject'].count())
# 3. Find maximum marks per subject.
print('\nFind maximum marks per subject\n', df.groupby(by = 'Subject')['Marks'].max())
```

## Task 9:  Mini Use Case: Sales Data Analysis

```python
''' 
Create a DataFrame:
sales = {
'Day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
'Revenue': [1200, 1500, 900, 2000, 1800]
}
Perform:
1. Total revenue
2. Average daily revenue
3. Day with highest revenue
4. Days where revenue > average
5. Plot revenue vs day
'''
sales = {
'Day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
'Revenue': [1200, 1500, 900, 2000, 1800]
}
df = pd.DataFrame(sales)
# 1. Total revenue
print('\nTotal revenue', df['Revenue'].sum())

# 2. Average daily revenue
avg_rev = df['Revenue'].mean()
print('\nAverage daily revenue', avg_rev)

# 3. Day with highest revenue
print('\nAverage daily revenue\n', df[df['Revenue']==df['Revenue'].max()])

# 4. Days where revenue > average
print('\nDays where revenue > average\n', df[df['Revenue']>avg_rev])

# 5. Plot revenue vs day
df.plot(x='Day', y='Revenue', xlabel='Day', ylabel='Revenue', title='Revenue vs Day', kind='bar')
df.plot(x='Day', y='Revenue', xlabel='Day', ylabel='Revenue', title='Revenue vs Day', )
```
