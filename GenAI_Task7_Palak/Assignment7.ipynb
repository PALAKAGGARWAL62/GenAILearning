## Task 1: Basic Class & Object Creation

```python
'''
Create a class Product with:
Attributes: name, price, category
Method: get_info() → prints product details
Create two objects and call get_info().
Extra (optional): Add a method apply_discount(percent) that returns the discounted price.
'''

# Class definition
class Product:
    name = ''
    price = 0
    category = ''
    def get_info(self,):
        # product information
        print('Name: %s, Price: %d, Category: %s'%(self.name, self.price, self.category))

    def apply_discount(self, percent):
        # discounted price 
        return (self.price)*(100 - percent)/100

# call object creation
product1 = Product()
product1.name = 'Apple'
product1.price = 200
product1.category = 'FnB'
product1.get_info()
dp = product1.apply_discount(10)
print("Discounted Price - ",dp)

# call object creation
product2 = Product()
product2.name = 'Bread'
product2.price = 100
product2.category = 'FnB'
product2.get_info()
dp = product2.apply_discount(5)
print("Discounted Price - ",dp)
```

## Task 2: Constructor & Encapsulation

```python
'''
Modify the Product class:
Make price a private attribute (__price).
Create getter & setter methods:
o get_price()
o set_price(new_price) → should update only if new_price > 0
Test modifying price using the setter.
'''

# Class definition
class Product:
    name = ''
    category = ''

    def __init__(self):
        self.__price = 0

    def get_price(self):
        return self.__price

    def set_price(self, new_price):
        if new_price>0:
            self.__price = new_price

    def get_info(self,):
        # product information
        print('Name: %s, Price: %d, Category: %s'%(self.name, self.__price, self.category))

    def apply_discount(self, percent):
        # discounted price 
        # getter for price
        return (self.get_price())*(100 - percent)/100

# call object creation
product1 = Product()
product1.name = 'Apple'
product1.price = 200 # unable to assign value as variable is private
product1.category = 'FnB'
product1.get_info()

product1.set_price(200) # using price setter function
print("Product Price: ", product1.get_price()) # getter function
product1.get_info()
dp = product1.apply_discount(10)
print("Discounted Price - ",dp)
```

## Task 3: Inheritance (Single-Level)

```python
'''
Create a subclass ElectronicProduct that inherits from Product.
Additional attribute:
warranty_years
Override the get_info() method to include warranty info.
Create an object and demonstrate inheritance + overriding.
'''

class ElectronicProduct(Product):
    warranty_years = 0
    category = 'Electronic'

    # overridding get_info function
    def get_info(self):
        print('Name: %s, Price: %d, Category: %s Warranty Years: %d'%(self.name, self.get_price(), self.category, self.warranty_years))


# call object creation
product1 = ElectronicProduct()
product1.name = 'Airpod'
# property of parent class
product1.set_price(2000) # using price setter function
# property of child class
product1.warranty_years = 10
# property of child class
product1.get_info()
# property of parent class
dp = product1.apply_discount(10)
print("Discounted Price - ",dp)
```

## Task 4: Polymorphism

```python
''' 
Create two classes:
Laptop(Product)
Mobile(Product)
Both override:
def get_info(self):
Print details in their own style
Write a loop that iterates over objects of Laptop and Mobile and calls get_info() on each to
show polymorphism.
'''

class Laptop(Product):
    # overridding get_info function
    def get_info(self):
        print('This is the Laptop child class - Name: %s, Price: %d, Category: %s'%(self.name, self.get_price(), self.category))

class Mobile(Product):
    # overridding get_info function
    def get_info(self):
        print('This is the Mobile child class - Name: %s, Price: %d, Category: %s'%(self.name, self.get_price(), self.category))

data = {'Laptop':[['Apple', '300000', 'Electornics'], ['Asus', '100000', 'Electornics'], ['Dell', '150000', 'Electornics']],
    'Mobile':[['Apple Phone', '150000', 'Electornics'], ['Android  Phone', '100000', 'Electornics'], ['Samsung', '140000', 'Electornics']]}
for k, v in data.items():
    for x in v:
        if k =='Laptop':
            obj = Laptop()
        if k =='Mobile':
            obj = Mobile()
        obj.name = x[0]
        obj.set_price(int(x[1]))
        obj.category = x[-1]
        obj.get_info()
```

## Task 5: Abstraction (Using Abstract Base Class)

```python
'''
Create an abstract class Payment with abstract method:
process_payment(amount)
Then create two subclasses:
CreditCardPayment
UPIPayment
Both override process_payment() with simple print statements.
Test all classes.
'''
from abc import ABC, abstractmethod

# Abstract Class
class Payment(ABC):
    @abstractmethod
    def process_payment(self, amount) -> None:
        pass

# child class
class CreditCardPayment(Payment):
    def process_payment(self, amount):
        print("Credit Card Payment class - ",amount )
        return super().process_payment(amount)

# child class
class UPIPayment(Payment):
    def process_payment(self, amount):
        print("UPI Payment class - ", amount)
        return super().process_payment(amount)

# Abstract class can not be instanciated, below 2 lines will give error
# obj = Payment()
# obj.process_payment(100)

# objects of child class
obj = CreditCardPayment()
obj.process_payment(200)

obj = UPIPayment()
obj.process_payment(300)
```

## Task 6: Magic Methods & Operator Overloading

```python
'''
Add the following to your Product class:
1. __str__ method
Returns a readable string:
Product(name, price, category)
2. Operator Overloading (__add__)
Allow:
product1 + product2
To return the total combined price.
Test this with two product objects.
'''

# Class definition
class Product:
    name = ''
    category = ''

    def __init__(self, *args):
        self.__price = args[-1]
        self.name = args[0]
        self.category = args[1]

    def __str__(self):
        # string representation of objects of the class is obtained by __str__ method
        return('Name: %s, Price: %d, Category: %s'%(self.name, self.__price, self.category))

    def __add__(self, prod):
        # magic function adding price of 2 different objects of class Product
        print(self.__price, prod.get_price(),)
        return Product('', '', self.__price+prod.get_price())
    
    def get_price(self):
        return self.__price

    def set_price(self, new_price):
        if new_price>0:
            self.__price = new_price

    def get_info(self,):
        # product information
        print('Name: %s, Price: %d, Category: %s'%(self.name, self.__price, self.category))

    def apply_discount(self, percent):
        # discounted price 
        # getter for price
        return (self.get_price())*(100 - percent)/100

# call object creation
product1 = Product('Apple', 'FnB', 300)
print(product1)
print('Type of Product object: ',type(product1)) # Type of object remains same but string is returned as its value
print(str(product1))
print('Type casting object: ', type(str(product1)))
# print(product1[0:3]) #slicing can  not be done on product1 as it is just object representation, to perform slicing type cast it to string

# usage of __add__
product2 = Product('Bread', 'Bakery', 100)
print(product2)
product3 = product1+product2
print(product3)
print('Sum of price: ', product3.get_price())
```

## Task 7: Mini Project: Simple Inventory System (OOP Only)

```python
''' 
Create two classes:
Class: Inventory
Attributes:
products → list to store product objects
Methods:
add_product(product)
remove_product(name)
get_total_value() → sums prices of all products
show_all_products() → prints info for each product
Class: Store
Attributes:
store_name
inventory → an Inventory object
Methods:
add_new_product() → takes input & creates Product object
show_summary() → prints total items & value
Important: Use only OOP concepts — no file handling, no exceptions, no packages.

Test the system by:
1. Creating a Store object
2. Adding 3 products
3. Showing summary
4. Using __add__ to combine prices of two products

Product class is added for managing products
product - name, price, category
'''

# Class definition
class Product:
    name = ''
    category = ''

    def __init__(self, *args):
        print(*args)
        self.__price = args[-1]
        self.name = args[0]
        self.category = args[1]

    def __str__(self):
        # string representation of objects of the class is obtained by __str__ method
        return('Name: %s, Price: %d, Category: %s'%(self.name, self.__price, self.category))

    def __add__(self, prod):
        # magic function adding price of 2 different objects of class Product
        print(self.__price, prod.get_price(),)
        return Product('', '', self.__price+prod.get_price())
    
    def get_price(self):
        return self.__price

    def set_price(self, new_price):
        if new_price>0:
            self.__price = new_price

    def get_info(self,):
        # product information
        print('Name: %s, Price: %d, Category: %s'%(self.name, self.__price, self.category))


class Inventory:
    products = []
    name = ''
    price = 0
    category = ''

    def __init__(self):
        pass

    def add_product(self, product):
        # create product
        self.products.append(product)
        return product

    def remove_product(self, name):
        # remove the product object from object using name
        for i in range(len(self.products)):
            if self.products[i].name==name:
                removedobj = self.products.pop(i)
                break
        return removedobj
    
    def get_total_value(self,):
        # sum of all the products price
        total = 0
        itms = len(self.products)
        for i in range(itms):
            total += self.products[i].get_price()
        return total, itms

    def show_all_products(self,):
        # prints information of each product
        for i in range(len(self.products)):
            self.products[i].get_info()
        return self.products

class Store():
    store_name = 'Smart Grocery Store'
    inventory = Inventory()
    def add_new_product(self,):
        # takes input to create the product object
        name = input('Enter the product name: ')
        price = int(input('Enter the price of the product: '))
        category = input("Enter the category of the product: ")
        product = Product(name, category, price,)
        # calling add_product method of inventory class to create products list 
        self.inventory.add_product(product)
        return product
    
    def show_summary(self,):
        # prints total value and items
        # display all products of inventory
        print('Summary of the store')
        self.inventory.show_all_products()
        # calculation of the summary
        total, itms = self.inventory.get_total_value()
        print('Total Items: ', total ,'\nTotal Value: ',itms)
        return total, itms

store_obj = Store()
print(store_obj.store_name)
n = int(input('Enter number of products: '))
for i in range(n):
    store_obj.add_new_product()
store_obj.show_summary()
if n>=2:
    print(store_obj.inventory.products[0]+store_obj.inventory.products[1])
```
