## Task 1: Safe Division Utility

```python
'''
Write a Program 
1. Takes 2 inputs from the user: numerator and denominator
2. Uses try-except to handle:
 - ValueError (if input is not a number)
 - ZeroDivisionError (if denominator = 0)
3.If no error occurs, print the result inside the else block
4. In finally block print('Operation Complete')
'''
div = 0
try:
    numerator = int(input('Enter numerator: '))
    denominator = int(input('Enter denominator: '))
    div = numerator/denominator
except ValueError as e:
    print('Input is not a number. It is a ValueError: ',e)
except ZeroDivisionError as e:
    print('Denominator is 0. It is a ZeroDivisionError: ', e)
else:
    print('%d / %d = %.2f '%(numerator, denominator, div))
finally:
    print("Operation Complete")
```

## Task 2: Bill Calculator with Error Handling

```python
'''
Given a list of product prices:
prices = [120, 350, 'abc', 500, -200, 800]
Write a Program 
1. Iterate through the list
2. Tries to add only valid(+ive num) prices to the total
3. Handles
- TypeError if value is not a number
- Custom exeception using raise ValueError ('Negative price not allowed')
4. Prints the running Total.
Expected Behaviour: 
Skip invalid items but continue processing
'''
prices = [120, 350, 'abc', 500, -200, 800]
total = 0
for x in prices:
    try:
        print('Price - ',x)
        try:
            if x<0:
                raise ValueError('Negative price not allowed')
        except ValueError as e:
            print('Custom ValueError: ',e)
            continue
        except TypeError as e:
            print("Value is not a number, TypeError: ",e)
            continue
        total+=x
    except Exception as e:
        print("Exception: ",e)
    print('Running Total - ', total)

```

##  Task 3: Custom Exception: Age Validator

```python
'''
1. Write function to check_age(age) that:
- Raise a custom Exception ValueError("Age must be between 1 and 120) if age is out of range
2. In your main code:
- Take age input from the user
- Use try-except to catch and print the custom error message
'''
def check_age(age):
    if not (age>=1 and age<=120):
        raise ValueError("Age must be between 1 and 120")
    else:
        return age

try:
    age = int(input('Enter age: '))
    age = check_age(age)
except ValueError as e:
    print('This is a ValueError: ', e)
else:
    print('Age: ', age)
```

## Task 4: File Reader with Exception Handling

```python
'''
1. Ask the user for a filename
2. Try to open and read the file
3. Handle:
- FileNotFoundError
- PermissionError
4. If successful, print first 3 lines of the file
5. Use finally to print - 'File operation attempted'
'''
import os
fname = input('Enter filename: ')
if '.' in fname:
    try:
        f = open(fname, 'r+')
        data = f.readlines()[:3]
        data = list(map(lambda x: x.strip(), data))
        for x in data:
            print(x)
        f.close()
    except FileNotFoundError as e:
        print('File doesn\'t exist. FileNotFoundError: ',e)
    except PermissionError as e:
        print('File PermissionError: ',e)
    finally:
        print('\nFile operation attempted')
else:
    print('Invalid File name')
```

## Task 5: Mini Program: Safe Shopping Cart

```python
"""
Create a program
1. Has cart list -> cart = []
2. Runs a loop asking user to enter prices
3. stops when user enters 'q'
4. Inside the loop:
- Convert input to a float
- Handle ValueError if user enters invalid input
- Raise custom exception if price is negative
5. At it end print:
- Total items
- Total bill
"""

cart = ['Apple', 'Bread', 'Butter', 'Milk', 'Eggs', 'Biscuits', 'Tea', 'Coffee']
price = ''
i = 0
total = 0
while i<len(cart) and price not in ['q', 'Q']:
    try:
        price = input('Enter Price of %s (Press q to quit): '%cart[i])
        if price in ['q', 'Q']:
            break
        # price conversion to float
        price = float(price)
        if price<0:
            # custom Exception
            raise ValueError('Price is negative')
        else:
            # adding up price for total bill
            total+=price
    except ValueError as e:
        print("Value Error: ", e)
    finally:
        i+=1

print('Total Items: ', i)
print('Total Bill: %.2f'%total)
```
