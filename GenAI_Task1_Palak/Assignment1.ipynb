# Task 3: Product Pricing (Dictionaries)

```python
# dictionary with product and price
price_dict = {'Sugar':70, 'Salt':30, 'Bread':60, 'Butter':60, 'Green Tea':100, 'Coffee':100}
print('Price Dictionary - ', price_dict)

# adding product to the price_dict
price_dict['Shampoo'] = 900
print('Adding Product to the price dictionary: ', price_dict)

# update existing product price
price_dict['Bread'] = 100
print('Updated the price of the bread: ', price_dict)

# remove product that do not exist
# removal of non existing product will give key error, to handle that we can apply conditions or exception handling
try:
    price_dict.pop('Oil')
except Exception as e:
    print("Exception while removing non existing product: ",e)
    print('Removed oil from the dictionary: ', price_dict)

# remove product that exists
price_dict.pop('Bread')
print('Removed Bread from the dictionary: ', price_dict)

# Average price of all the products
counter = 0
price_sum = 0
max_price = 0
min_price = 0
for k in price_dict:
    price_sum += price_dict[k]
    counter += 1
    if counter==1:
        max_price = price_dict[k]
        min_price = price_dict[k]
    if max_price < price_dict[k]:
        max_price = price_dict[k]
    if min_price > price_dict[k]:
        min_price = price_dict[k]

price_avg = price_sum / counter
print("Price average : ", price_avg, "\nMaximum Price: ", max_price, '\nMinimum Price: ',min_price, '\nProduct of Maximum and Minimum Price: ', max_price*min_price)

```

# Task 4: Combined Operations

```python
# using products list, price_dict, create a list of tuples (product_name, price, category)
product = ['Sugar', 'Salt', 'Bread', 'Butter', 'Green Tea', 'Coffee', 'Shampoo', 'Oil', 'Avocado']
price_dict = {'Sugar': 70, 'Salt': 30, 'Bread': 60, 'Butter': 60, 'Green Tea': 100, 'Coffee': 100, 'Shampoo': 900, 'Oil':200, 'Avocado':200}
categories = ['Grocery', 'Grocery', 'F&B', 'F&B', 'Grocery', 'Grocery', 'Beauty', 'Beauty', 'F&B', ]
catalog = []
for i in range(len(product)):
    catalog.append((product[i], price_dict[product[i]], categories[i]))
print("Product Catalog")
print(catalog)

# from catalog create a new dict category_to_products that maps each category to a list of product names in the category

category_to_products = {}
for row in catalog:
    if row[-1] not in category_to_products: 
        category_to_products[row[-1]] = [row[0]]
    else:
        category_to_products[row[-1]].append(row[0])

# print the products of the category having max products
counter = 0
max_product = []
max_category = ''
for k, v in category_to_products.items():
    if len(v)>counter:
        counter = len(v)
        max_product = v
        max_category = k

print('Category with maximum products is : ', max_category, ': ', max_product)
```
